```python
import random
import time

def slow_print(text, delay=0.03):
    for char in text:
        print(char, end='', flush=True)
        time.sleep(delay)
    print()

def start_game():
    player_hp = 100
    gold = 0
    inventory = []

    slow_print("=== 🌲 안개 낀 미궁 숲 탐험기 🌲 ===")
    slow_print("당신은 깊은 숲속의 고대 유적 앞에 서 있습니다.")
    
    while player_hp > 0:
        print(f"\n[현재 상태] HP: {player_hp} | 골드: {gold}G | 소지품: {inventory}")
        slow_print("\n어디로 이동하시겠습니까?")
        print("1. 어두운 동굴 깊은 곳으로 들어간다.")
        print("2. 빛이 새어나오는 폐허를 수색한다.")
        print("3. 휴식을 취하며 체력을 회복한다.")
        print("4. 탐험을 종료하고 마을로 귀환한다.")
        
        choice = input("선택 (1~4): ").strip()
        
        if choice == "1":
            slow_print("\n동굴 안에서 붉은 눈의 괴물이 나타났습니다!")
            event = random.choice(["monster", "chest", "trap"])
            
            if event == "monster":
                monster_hp = 40
                while monster_hp > 0 and player_hp > 0:
                    damage_dealt = random.randint(15, 25)
                    damage_taken = random.randint(5, 15)
                    monster_hp -= damage_dealt
                    player_hp -= damage_taken
                    slow_print(f"공격! 괴물에게 {damage_dealt}의 피해를 입혔습니다.")
                    if monster_hp > 0:
                        slow_print(f"괴물의 반격! {damage_taken}의 피해를 입었습니다.")
                
                if player_hp > 0:
                    reward = random.randint(20, 50)
                    gold += reward
                    slow_print(f"괴물을 물리쳤습니다! 전리품으로 {reward}G를 획득했습니다.")
            elif event == "chest":
                slow_print("괴물 대신 숨겨진 보물상자를 발견했습니다! 50G를 얻었습니다.")
                gold += 50
            else:
                slow_print("함정을 밟아 화살을 맞았습니다! 체력 -20")
                player_hp -= 20

        elif choice == "2":
            slow_print("\n폐허의 잔해 사이를 조심스럽게 탐색합니다...")
            find_item = random.choice(["포션", "고대 유물", "꽝"])
            if find_item == "포션":
                slow_print("생명력의 물약을 찾았습니다! 인벤토리에 보관합니다.")
                inventory.append("HP 물약")
            elif find_item == "고대 유물":
                slow_print("반짝이는 에테르 원석을 발견했습니다! 가치가 높아 보입니다.")
                inventory.append("에테르 원석")
                gold += 80
            else:
                slow_print("쓸만한 물건은 없었습니다. 모래바람만 날립니다.")

        elif choice == "3":
            if "HP 물약" in inventory:
                inventory.remove("HP 물약")
                player_hp = min(100, player_hp + 40)
                slow_print("\nHP 물약을 마셨습니다! 체력이 40 회복되었습니다.")
            else:
                heal = random.randint(10, 20)
                player_hp = min(100, player_hp + heal)
                slow_print(f"\n모닥불 옆에서 푹 쉬었습니다. 체력이 {heal} 회복되었습니다.")

        elif choice == "4":
            slow_print(f"\n탐험을 마치고 무사히 탈출했습니다!")
            slow_print(f"최종 결과 -> 남은 체력: {player_hp} | 최종 골드: {gold}G | 수집 아이템: {inventory}")
            break
        else:
            print("잘못된 입력입니다. 1~4 중 선택해 주세요.")

    if player_hp <= 0:
        slow_print("\n체력이 0이 되었습니다... 숲의 어둠 속으로 쓰러집니다. [GAME OVER]")

if __name__ == "__main__":
    start_game()
```
